```python
# Load the data
from langchain_community.document_loaders import WebBaseLoader

# List of 10 relevant Wikipedia articles
wiki_urls = [
    "https://en.wikipedia.org/wiki/Artificial_intelligence",
    "https://en.wikipedia.org/wiki/Machine_learning",
    "https://en.wikipedia.org/wiki/Deep_learning",
    "https://en.wikipedia.org/wiki/Natural_language_processing",
    "https://en.wikipedia.org/wiki/Computer_vision",
    "https://en.wikipedia.org/wiki/Artificial_neural_network",
    "https://en.wikipedia.org/wiki/Large_language_model",
    "https://en.wikipedia.org/wiki/Generative_adversarial_network",
    "https://en.wikipedia.org/wiki/Reinforcement_learning",
    "https://en.wikipedia.org/wiki/Ethics_of_artificial_intelligence"
]

# Initialize the loader with the list of URLs
loader = WebBaseLoader(wiki_urls)

# Load all documents
docs = loader.load()

# Inspect the loaded documents
print(f"Loaded {len(docs)} documents.")
print("\n--- Sample Document (from the first URL) ---")
print(docs[0].page_content[:50]) # Print the first 50 characters
```

```python
# Fixed size Chunks
from langchain_text_splitters import CharacterTextSplitter

text_splitter_fixed = CharacterTextSplitter(
    separator = "",
    chunk_size = 1000,
    chunk_overlap = 100,
    length_function = len,
)

fixed_chunks = text_splitter_fixed.split_documents(docs)
print(f"Created {len(fixed_chunks)} fixed-size chunks.")
print(f"Sample chunk: \n{fixed_chunks[0].page_content}")
```

```python
# Recursive based chunks
from langchain_text_splitters import RecursiveCharacterTextSplitter
text_splitter_recursive = RecursiveCharacterTextSplitter(
    chunk_size = 1000,
    chunk_overlap = 100,
   add_start_index=True,
)

recursive_chunks = text_splitter_recursive.split_documents(docs)
print(f"Created {len(recursive_chunks)} recursive-size chunks.")
print(f"Sample chunk: \n{recursive_chunks[0].metadata}")
```

```python
# Semantic based chunks
from langchain_experimental.text_splitter import SemanticChunker
from langchain_google_genai import GoogleGenerativeAIEmbeddings

embeddings_for_semantic = GoogleGenerativeAIEmbeddings(model = "models/gemini-embedding-001",google_api_key=userdata.get('GOOGLE_API_KEY') )

text_splitter_semantic = SemanticChunker(embeddings_for_semantic)

# for single document only
semantic_chunks = text_splitter_semantic.create_documents([docs[0].page_content])
print(f"Created {len(semantic_chunks)} semantic chunks from the first document.")
print(f"Sample chunk: \n{semantic_chunks[1].page_content}")
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.output_parsers import JsonOutputParser
from langchain_core.prompts import ChatPromptTemplate

llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash", temperature=0)

prompt = ChatPromptTemplate.from_template(
    "You are an expert at splitting long documents into smaller, contextually-aware sections.\n"
    "Split the following document into a list of sections. Each section should be a self-contained piece of text with relevant context.\n"
    "Return a JSON object with a single key 'chunks' containing a list of strings.\n"
    "Document: {document}"
)

chunking_chain = prompt | llm | JsonOutputParser()

# Let's use a small slice of a document to demonstrate
document_slice = docs[1].page_content[0:3000]
llm_chunks_json = chunking_chain.invoke({"document": document_slice})

print(f"LLM created {len(llm_chunks_json['chunks'])} chunks.")
print(f"Sample chunk: \n{llm_chunks_json['chunks'][0]}")
```

```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings

# Initialize the embedding model
embeddings = GoogleGenerativeAIEmbeddings(model="models/gemini-embedding-001")
```

```python
#create chroma db
from langchain_chroma import Chroma
vector_store = Chroma(
    embedding_function = embeddings,
    persist_directory = "./chroma_db_splits"
)
```

```python
batch_size = 100
chunks = recursive_chunks
import time
print("========== Starting to add documents in DB =============== \n")

for i in range(0,len(chunks),batch_size):
  batch = chunks[i:i+batch_size]
  vector_store.add_documents(documents=batch)
  # Print progress
  print(f"Added batch {i // batch_size + 1} of {(len(chunks) - 1) // batch_size + 1}")
  # Pause for a moment to avoid hitting the rate limit for the free tier
  time.sleep(10)
print("======== Documents added in DB Successfully ============== \n")
```
